# Resolver o quebra-cabeças usando Buscas

## Busca em largura

```python
import copy
import heapq

estado_inicial =  [[1,0,3],[4,5,6],[7,8,2]]
estado_final =  [[1,2,3],[4,5,6],[7,8,0]]
movimentos = [(-1, 0), (1, 0), (0, -1), (0, 1)]
explorados = set()


class No:
  def __init__(self, estado_atual, acao, pai, custo_do_caminho):
      self.estado_atual = estado_atual
      self.acao = acao
      self.pai = pai
      self.custo_do_caminho = custo_do_caminho



def abrir_No(no_pai, movimentos):

  vizinhos = []

  for linha_idx, linha in enumerate(no_pai.estado_atual):
    if 0 in linha:
      coluna_idx = linha.index(0)
      posicao_x, posicao_y = linha_idx, coluna_idx
      break

  for movimento in movimentos:

    if(0 <= movimento[0] + posicao_x <= 2 and 0 <= movimento[1] + posicao_y <= 2):

      novo_estado = copy.deepcopy(no_pai.estado_atual)

      novo_x = posicao_x + movimento[0]
      novo_y = posicao_y + movimento[1]

      novo_estado[posicao_x][posicao_y], novo_estado[novo_x][novo_y] = novo_estado[novo_x][novo_y], novo_estado[posicao_x][posicao_y]

      filho = No(novo_estado, movimento, no_pai, no_pai.custo_do_caminho + 1)

      vizinhos.append(filho)

  return vizinhos



def BFS(tabuleiro):

  inicio = No(tabuleiro, None, None, 0)

  fronteira = [inicio]

  while(len(fronteira) > 0):

   no_atual = fronteira.pop(0)

   if(no_atual.estado_atual == estado_final):

    return no_atual

   else:

    if(str(no_atual.estado_atual) not in explorados):

      explorados.add(str(no_atual.estado_atual))
      fronteira.extend(abrir_No(no_atual, movimentos))




vencedor = BFS(estado_inicial)

no = vencedor
caminho = []
direcoes = []

while (no != None):

  caminho.append(no.estado_atual)
  direcoes.append(no.acao)

  no = no.pai

passos = len(caminho) - 1

print(passos)
print(len(explorados)-1)
for tabuleiro, direcao in zip(caminho[::-1], direcoes[::-1]):

    if direcao is not None:
        print(direcao)

    for linha in tabuleiro:
        print(linha)

    print("-" * 15)

```

## Busca em profundidade com aprofundamento iterativo

```python
import copy
import heapq

estado_inicial =  [[1,0,3],[4,5,6],[7,8,2]]
estado_final =  [[1,2,3],[4,5,6],[7,8,0]]
movimentos = [(-1, 0), (1, 0), (0, -1), (0, 1)]



class No:
  def __init__(self, estado_atual, acao, pai, custo_do_caminho):
      self.estado_atual = estado_atual
      self.acao = acao
      self.pai = pai
      self.custo_do_caminho = custo_do_caminho



def abrir_No(no_pai, movimentos):

  vizinhos = []

  for linha_idx, linha in enumerate(no_pai.estado_atual):
    if 0 in linha:
      coluna_idx = linha.index(0)
      posicao_x, posicao_y = linha_idx, coluna_idx
      break

  for movimento in movimentos:

    if(0 <= movimento[0] + posicao_x <= 2 and 0 <= movimento[1] + posicao_y <= 2):

      novo_estado = copy.deepcopy(no_pai.estado_atual)

      novo_x = posicao_x + movimento[0]
      novo_y = posicao_y + movimento[1]

      novo_estado[posicao_x][posicao_y], novo_estado[novo_x][novo_y] = novo_estado[novo_x][novo_y], novo_estado[posicao_x][posicao_y]

      filho = No(novo_estado, movimento, no_pai, no_pai.custo_do_caminho + 1)

      vizinhos.append(filho)

  return vizinhos



def IDDFS(tabuleiro):

  for limite in range(1, 100):

    inicio = No(tabuleiro, None, None, 0)

    explorados = {}

    fronteira = [inicio]

    while(len(fronteira) > 0):

      no_atual = fronteira.pop()

      if(no_atual.estado_atual == estado_final):

        return no_atual

      else:

        if(no_atual.custo_do_caminho < limite):

          atual = str(no_atual.estado_atual)
          if (atual not in explorados) or (no_atual.custo_do_caminho < explorados[atual]):

            explorados[atual] = no_atual.custo_do_caminho

            fronteira.extend(abrir_No(no_atual, movimentos))



vencedor = IDDFS(estado_inicial)

no = vencedor
caminho = []
direcoes = []

while (no != None):

  caminho.append(no.estado_atual)
  direcoes.append(no.acao)

  no = no.pai

passos = len(caminho) - 1

print(passos)
print(len(explorados)-1)

for tabuleiro, direcao in zip(caminho[::-1], direcoes[::-1]):

    if direcao is not None:
        print(direcao)

    for linha in tabuleiro:
        print(linha)

    print("-" * 15)

```

## Discorra sobre o desempenho dos métodos em questões de:


1.   Consumo de memória
2.   Processamento



### **A conclusão que tirei é a de que o DFS, mesmo demorando muito mais tempo de processamento, consome, dadas as devidas proporções, menos memória que o BFS, que, apesar de rápido, consome muita memória. Já que o DFS não guarda a informação de todos os vizinhos assim como o BFS**

### **No IDDFS aconteceu uma coisa curiosa, ele tem o mesmo numero de caminhos explorados do DFS porém o numero de passos é absurdamente menor, a primeira vez que implementei ele estava com 15 passos, até eu entender que ele nao estava encontrando o menor numero de passos como o bfs pois ele ignorava caminhos já vistos mesmo que o caminho repetido agora tenha um custo menor que o atual, fazendo com que o resultado do caminho final nao fosse o melhor**

## Segue o código abaixo que usei para fazer a comparação:






```python
import time
import tracemalloc

tracemalloc.start()
tempo_inicio_bfs = time.time()

puzzle_bfs = BFS(estado_inicial)

tempo_fim_bfs = time.time()
memoria_atual_bfs, memoria_pico_bfs = tracemalloc.get_traced_memory()
tracemalloc.stop()

tempo_total_bfs = tempo_fim_bfs - tempo_inicio_bfs
pico_mb_bfs = memoria_pico_bfs / (1024 * 1024)

print(f"Tempo de Processamento BFS: {tempo_total_bfs:.4f} segundos")
print(f"Pico de Memória Consumida BFS: {pico_mb_bfs:.4f} MB\n")


tracemalloc.start()
tempo_inicio_dfs = time.time()

puzzle_dfs = DFS(estado_inicial)

tempo_fim_dfs = time.time()
memoria_atual_dfs, memoria_pico_dfs = tracemalloc.get_traced_memory()
tracemalloc.stop()

tempo_total_dfs = tempo_fim_dfs - tempo_inicio_dfs
pico_mb_dfs = memoria_pico_dfs / (1024 * 1024)

print(f"Tempo de Processamento DFS: {tempo_total_dfs:.4f} segundos")
print(f"Pico de Memória Consumida DFS: {pico_mb_dfs:.4f} MB\n")


tracemalloc.start()
tempo_inicio_iddfs = time.time()

puzzle_dfs = IDDFS(estado_inicial)

tempo_fim_iddfs = time.time()
memoria_atual_iddfs, memoria_pico_iddfs = tracemalloc.get_traced_memory()
tracemalloc.stop()

tempo_total_iddfs = tempo_fim_iddfs - tempo_inicio_iddfs
pico_mb_iddfs = memoria_pico_iddfs / (1024 * 1024)

print(f"Tempo de Processamento IDDFS: {tempo_total_iddfs:.4f} segundos")
print(f"Pico de Memória Consumida IDDFS: {pico_mb_iddfs:.4f} MB")
```
